```python
import pandas as pd
import numpy as np

# Load dataset
file_path = '/kaggle/input/datasets/fahim173/stroke-risk/stroke_risk_prediction_dataset (1).csv'
df = pd.read_csv(file_path)

# Drop identifier column if present as it has no predictive power
if 'Patient_ID' in df.columns:
    df = df.drop(columns=['Patient_ID'])

# Define target variable and feature set
target_col = 'Stroke_Risk'
X = df.drop(columns=[target_col])
y = df[target_col]

print(f"Dataset Shape: {df.shape}")
print(f"Features: {X.shape[1]} columns | Target: {target_col}")
```

```python
# Calculate count and percentage of missing values
missing_count = X.isnull().sum()
missing_pct = (X.isnull().mean()) * 100

# Create summary DataFrame
missing_summary = pd.DataFrame({
    'Missing Count': missing_count,
    'Missing Percentage (%)': missing_pct
})

# Display features with missing values
print("--- Missing Values Summary ---")
print(missing_summary[missing_summary['Missing Count'] > 0].sort_values(by='Missing Percentage (%)', ascending=False))
```

```python
# Identify columns with > 20% missing values
high_missing_cols = missing_pct[missing_pct > 20].index.tolist()

print(f"Columns to drop (> 20% missing): {high_missing_cols}")

# Drop identified columns
X = X.drop(columns=high_missing_cols)
```

```python
# Identify columns with > 20% missing values
high_missing_cols = missing_pct[missing_pct > 20].index.tolist()

print(f"Columns to drop (> 20% missing): {high_missing_cols}")

# Drop identified columns
X = X.drop(columns=high_missing_cols)
```

```python
# Impute numerical features with median and categorical features with mode
for col in X.columns:
    if X[col].isnull().sum() > 0:
        if X[col].dtype in ['float64', 'int64']:
            X[col] = X[col].fillna(X[col].median())
        else:
            X[col] = X[col].fillna(X[col].mode()[0])

print("Imputation completed.")
```

```python
# Final verification of missing values
total_missing = X.isnull().sum().sum()
print(f"Total missing values remaining in features: {total_missing}")
```

```python
from sklearn.preprocessing import LabelEncoder

# Encode categorical feature columns
label_encoders = {}
X_encoded = X.copy()

categorical_cols = X_encoded.select_dtypes(include=['object', 'category']).columns

for col in categorical_cols:
    le = LabelEncoder()
    X_encoded[col] = le.fit_transform(X_encoded[col])
    label_encoders[col] = le

# Encode target variable
le_y = LabelEncoder()
y_encoded = le_y.fit_transform(y)

print("Label Encoding finished. Dataset is ready for model fitting.")
```

```python
from sklearn.model_selection import train_test_split

# 80/20 train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, 
    y_encoded, 
    test_size=0.20, 
    random_state=42, 
    stratify=y_encoded
)

print(f"X_train shape: {X_train.shape}")
print(f"X_test shape:  {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape:  {y_test.shape}")
```

```python
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Initialize and train Gaussian Naive Bayes classifier
nb_model = GaussianNB()
nb_model.fit(X_train, y_train)

# Predictions
y_pred = nb_model.predict(X_test)

# Evaluation metrics
accuracy = accuracy_score(y_test, y_pred)
print("Naive Bayes Accuracy:", round(accuracy * 100, 2), "%")

print("--- Classification Report ---")
print(classification_report(y_test, y_pred, target_names=le_y.classes_))

print("--- Confusion Matrix ---")
print(confusion_matrix(y_test, y_pred))
```

```python
# Merge features and encoded target to calculate Pearson correlation
df_encoded = X_encoded.copy()
df_encoded['Target'] = y_encoded

# Pearson correlation with target
pearson_corr = df_encoded.corr(method='pearson')['Target'].drop('Target').sort_values(ascending=False)

print("--- Pearson Correlation with Target (Top 10 Positively Correlated) ---")
print(pearson_corr.head(10))
```
